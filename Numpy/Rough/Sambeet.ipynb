```python
def dvide_numbers(x,y):
    try:
        result = x/y
        print(f'result is {result}')
    except ZeroDivisionError:
        print('Error: division by zero')
    except TypeError:
        print('Invalid input type')


dvide_numbers(10,0)
dvide_numbers(10,'a')
dvide_numbers(10,2)
```
